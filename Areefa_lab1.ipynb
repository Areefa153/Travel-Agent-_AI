```python
# Smart Travel Agent

print("================================")
print("       SMART TRAVEL AGENT")
print("================================")

name = input("Enter your name: ")

print("\nAvailable Destinations:")
print("1. Karachi")
print("2. Islamabad")
print("3. Murree")
print("4. Hunza")
print("5. Skardu")

destination = int(input("\nSelect destination (1-5): "))
days = int(input("Enter number of days: "))
budget = int(input("Enter your budget (PKR): "))

# Agent calculates package cost

if destination == 1:
    place = "Karachi"
    daily_cost = 5000

elif destination == 2:
    place = "Islamabad"
    daily_cost = 6500

elif destination == 3:
    place = "Murree"
    daily_cost = 7500

elif destination == 4:
    place = "Hunza"
    daily_cost = 9000

elif destination == 5:
    place = "Skardu"
    daily_cost = 11000

else:
    print("Invalid destination!")
    place = None

if place:

    total_cost = daily_cost * days

    print("\n---------- TRAVEL PLAN ----------")
    print("Traveler:", name)
    print("Destination:", place)
    print("Days:", days)
    print("Estimated Cost: PKR", total_cost)

    if budget >= total_cost:
        remaining = budget - total_cost
        print("Result: Package Available")
        print("Remaining Budget: PKR", remaining)
        print("Agent Recommendation: You can book this trip!")

    else:
        shortage = total_cost - budget
        print("Result: Budget Not Enough")
        print("Additional Amount Needed: PKR", shortage)
        print("Agent Recommendation: Increase your budget or reduce trip days.")

print("\nThank you for using Smart Travel Agent!")
```
